## Step 2: Configure the model

Meridian uses Bayesian framework and Markov Chain Monte Carlo (MCMC) algorithms to sample from the posterior distribution.

1\. Inititalize the `Meridian` class by passing the loaded data and the customized model specification. One advantage of Meridian lies in its capacity to calibrate the model directly through ROI priors, as described in [Media Mix Model Calibration With Bayesian Priors](https://research.google/pubs/media-mix-model-calibration-with-bayesian-priors/). In this particular example, the ROI priors for all media channels are identical, with each being represented as Lognormal(0.2, 0.9).

```python
import pandas as pd
import numpy as np

def extract_march_contributions(mmm):
    # 1. Dynamically find the dimension that represents time
    # We look for names like 'time' or 'date', or the one that isn't geo/media
    all_dims = list(mmm.input_data.media.dims)
    time_dim_name = next((d for d in all_dims if d in ['time', 'date', 'period']), None)

    if time_dim_name is None:
        # Fallback: find the dim that isn't 'geo' or 'media_channel'
        time_dim_name = next((d for d in all_dims if d not in ['geo', 'media_channel']), None)

    if time_dim_name is None:
        print(f"Error: Could not identify time dimension. Found: {all_dims}")
        return

    try:
        raw_dates = mmm.input_data.media.coords[time_dim_name].values
        dates = pd.to_datetime(raw_dates)
    except Exception as e:
        print(f"Error parsing dates from dimension '{time_dim_name}': {e}")
        return

    # 2. Identify March indices
    march_mask = (dates.month == 3)
    march_indices = np.where(march_mask)[0]

    if len(march_indices) == 0:
        print(f"Error: No March dates found in the '{time_dim_name}' dimension.")
        return

    # 3. Locate the Contribution Tensor
    posterior = mmm.inference_data.posterior
    # Common Meridian naming conventions for contribution
    possible_tensors = ['media_contribution', 'incremental_outcome', 'expected_media_contribution']
    tensor_name = next((name for name in possible_tensors if name in posterior), None)

    if tensor_name is None:
        print("Error: Contribution tensor not found. Ensure the model has finished sampling.")
        return

    contribution_tensor = posterior[tensor_name]

    # 4. Slice for March
    # This handles the slice correctly even if time is the 1st or 2nd dimension
    march_contributions = contribution_tensor.isel({time_dim_name: march_indices})

    # 5. Aggregate: Sum over Time, Mean over Chains/Draws, Sum over Geos
    # We sum 'geo' because this is a national report
    geo_dim_name = 'geo' if 'geo' in march_contributions.dims else march_contributions.dims[2]

    march_sum_posterior = march_contributions.sum(dim=[time_dim_name, geo_dim_name])
    march_mean_contributions = march_sum_posterior.mean(dim=['chain', 'draw']).values

    # 6. Final Output
    media_names = mmm.input_data.media.coords['media_channel'].values
    total_march_media_vol = np.sum(march_mean_contributions)

    print(f"--- Results for {tensor_name} ---")
    print(f"Total Modeled Media Volume for March: {total_march_media_vol:,.2f}\n")
    print("| Channel | March Modeled Volume | March Share % |")
    print("| :--- | :--- | :--- |")

    for i, name in enumerate(media_names):
        volume = march_mean_contributions[i]
        share = (volume / total_march_media_vol) * 100 if total_march_media_vol > 0 else 0
        print(f"| **{name}** | {volume:,.2f} | {share:.2f}% |")

# Execute
extract_march_contributions(mmm)
```

```python
import os
import pickle
import numpy as np
import pandas as pd

try:
    import cloudpickle
except ImportError:
    cloudpickle = None
try:
    import joblib
except ImportError:
    joblib = None
try:
    import arviz as az
except ImportError:
    az = None


def robust_load_model(path):
    """Defensively attempts to load the model file using multiple serialization protocols."""
    if not os.path.exists(path):
        raise FileNotFoundError(f"No file found at path: {path}")

    # Strategy 1: Standard Pickle
    try:
        with open(path, "rb") as f:
            return pickle.load(f)
    except Exception:
        pass

    # Strategy 2: Cloudpickle
    if cloudpickle:
        try:
            with open(path, "rb") as f:
                return cloudpickle.load(f)
        except Exception:
            pass

    # Strategy 3: Joblib
    if joblib:
        try:
            return joblib.load(path)
        except Exception:
            pass

    # Strategy 4: ArviZ NetCDF
    if az:
        try:
            return az.from_netcdf(path)
        except Exception:
            pass

    raise RuntimeError(
        f"Critical Failure: Unable to unpickle {path}. File structure matches no known serialization formats."
    )


def extract_may_shares_math(model_path, input_data, model_label="Model"):
    print(f"\n" + "=" * 60)
    print(f"Processing {model_label} via Tensor Math Reconstruction...")
    print("=" * 60)

    # 1. Load the model and isolate the posterior
    model = robust_load_model(model_path)
    idata = model.inference_data if hasattr(model, "inference_data") else model
    print("✅ Model parameters loaded successfully.")

    # 2. Extract Raw Tensors from Meridian InputData Object
    try:
        media_raw = np.array(input_data.media)
        time_coords = pd.to_datetime(np.array(input_data.time))

        if hasattr(input_data, 'media_channels'):
            channels = np.array(input_data.media_channels)
        else:
            channels = idata.posterior.coords['media_channel'].values

    except Exception as e:
        print(f"❌ Failed to extract raw tensors from your data object: {e}")
        return None

    # Handle Dimensionality: Force into (time, geo, channel)
    n_time = len(time_coords)

    if media_raw.ndim == 3:
        if media_raw.shape[1] == n_time:
            # Tensor is (geo, time, channel) -> transpose to (time, geo, channel)
            media_raw = np.transpose(media_raw, (1, 0, 2))
    elif media_raw.ndim == 2:
        if media_raw.shape[0] != n_time:
            # Tensor is (channel, time) -> transpose to (time, channel)
            media_raw = media_raw.T
        # Add a dummy geo dimension: (time, 1, channel)
        media_raw = media_raw[:, np.newaxis, :]

    print(f"✅ Tensors aligned successfully. Working shape: {media_raw.shape}")

    # 3. Create May 2026 Mask
    may_mask = (time_coords >= "2026-05-01") & (time_coords <= "2026-05-31")
    if not may_mask.any():
        print("❌ No time periods found for May 2026.")
        return None
    print(f"-> Isolated {may_mask.sum()} periods for May 2026.")

    # 4. Extract Posterior Means
    alpha = idata.posterior['alpha_m'].mean(dim=['chain', 'draw']).values
    ec = idata.posterior['ec_m'].mean(dim=['chain', 'draw']).values
    slope = idata.posterior['slope_m'].mean(dim=['chain', 'draw']).values

    if 'beta_gm' in idata.posterior:
        beta = idata.posterior['beta_gm'].mean(dim=['chain', 'draw']).values
    else:
        beta = idata.posterior['beta_m'].mean(dim=['chain', 'draw']).values
        beta = beta[np.newaxis, :]

    # 5. Scale the Media
    media_scaled = np.zeros_like(media_raw, dtype=float)
    for c in range(len(channels)):
        c_max = np.max(media_raw[:, :, c])
        if c_max > 0:
            media_scaled[:, :, c] = media_raw[:, :, c] / c_max

    # 6. Apply Adstock & Saturation
    time_steps, n_geos, n_channels = media_scaled.shape
    adstocked = np.zeros_like(media_scaled)

    for t in range(time_steps):
        if t == 0:
            adstocked[t] = media_scaled[t]
        else:
            adstocked[t] = media_scaled[t] + (alpha * adstocked[t-1])

    transformed = (adstocked ** slope) / ((adstocked ** slope) + (ec ** slope))

    # 7. Apply Beta Coefficients and Sum for May
    contributions = {}
    total_may_media = 0.0

    for c_idx, channel in enumerate(channels):
        channel_contrib = 0.0
        for g_idx in range(n_geos):
            effect = beta[g_idx, c_idx] * transformed[:, g_idx, c_idx]
            # Now `effect` and `may_mask` are guaranteed to both be length 124
            channel_contrib += effect[may_mask].sum()

        contributions[str(channel)] = max(0, channel_contrib)
        total_may_media += max(0, channel_contrib)

    # 8. Calculate Percentages
    if total_may_media == 0:
        print("⚠️ Total media contribution calculated as 0.")
        return None

    share_table = []
    for ch, vol in contributions.items():
        share_table.append({
            "Channel": ch,
            "May 2026 Contribution %": (vol / total_may_media) * 100
        })

    return pd.DataFrame(share_table).sort_values(by="May 2026 Contribution %", ascending=False)

# =====================================================================
# --- RUN PIPELINE ---
# =====================================================================

try:
    leads_path = "/content/model_leads.pkl"
    leads_shares = extract_may_shares_math(leads_path, data, "Leads Funnel")

    if leads_shares is not None:
        print("\n" + "🏆" * 25)
        print("🏆 MAY 2026 IN-SAMPLE CONTRIBUTION SHARES: LEADS")
        print("🏆" * 25)
        print(leads_shares.to_string(index=False, float_format="%.2f%%"))

except Exception as err:
    print(f"\n❌ Execution stopped for Leads Funnel: {err}")

try:
    apps_path = "/content/model_applications.pkl"
    apps_shares = extract_may_shares_math(apps_path, data, "Applications Funnel")

    if apps_shares is not None:
        print("\n" + "🏆" * 25)
        print("🏆 MAY 2026 IN-SAMPLE CONTRIBUTION SHARES: APPLICATIONS")
        print("🏆" * 25)
        print(apps_shares.to_string(index=False, float_format="%.2f%%"))

except Exception as err:
    print(f"\n❌ Execution stopped for Applications Funnel: {err}")
```

```python
#adding organic

import os
import pickle
import numpy as np
import pandas as pd

try:
    import cloudpickle
except ImportError:
    cloudpickle = None
try:
    import joblib
except ImportError:
    joblib = None
try:
    import arviz as az
except ImportError:
    az = None


def robust_load_model(path):
    """Defensively attempts to load the model file using multiple serialization protocols."""
    if not os.path.exists(path):
        raise FileNotFoundError(f"No file found at path: {path}")

    # Strategy 1: Standard Pickle
    try:
        with open(path, "rb") as f:
            return pickle.load(f)
    except Exception:
        pass

    # Strategy 2: Cloudpickle
    if cloudpickle:
        try:
            with open(path, "rb") as f:
                return cloudpickle.load(f)
        except Exception:
            pass

    # Strategy 3: Joblib
    if joblib:
        try:
            return joblib.load(path)
        except Exception:
            pass

    # Strategy 4: ArviZ NetCDF
    if az:
        try:
            return az.from_netcdf(path)
        except Exception:
            pass

    raise RuntimeError(
        f"Critical Failure: Unable to unpickle {path}. File structure matches no known serialization formats."
    )


def extract_may_shares_math(model_path, input_data, model_label="Model"):
    print(f"\n" + "=" * 60)
    print(f"Processing {model_label} via Tensor Math Reconstruction...")
    print("=" * 60)

    # 1. Load the model and isolate the posterior
    model = robust_load_model(model_path)
    idata = model.inference_data if hasattr(model, "inference_data") else model
    print("✅ Model parameters loaded successfully.")

    # 2. Extract Time Mask
    try:
        time_coords = pd.to_datetime(np.array(input_data.time))
        n_time = len(time_coords)
        may_mask = (time_coords >= "2026-05-01") & (time_coords <= "2026-05-31")
        if not may_mask.any():
            print("❌ No time periods found for May 2026.")
            return None
        print(f"-> Isolated {may_mask.sum()} periods for May 2026.")
    except Exception as e:
        print(f"❌ Failed to extract time coordinates: {e}")
        return None

    # --- Nested Helper Function for Tensor Math ---
    def calculate_tensor_effects(media_raw, channels, alpha, ec, slope, beta):
        """Applies Meridian transformations and sums effects for the May mask."""
        working_media = media_raw.copy()

        # Handle Dimensionality: Force into (time, geo, channel)
        if working_media.ndim == 3:
            if working_media.shape[1] == n_time:
                working_media = np.transpose(working_media, (1, 0, 2))
        elif working_media.ndim == 2:
            if working_media.shape[0] != n_time:
                working_media = working_media.T
            working_media = working_media[:, np.newaxis, :]

        # Scale the Media
        media_scaled = np.zeros_like(working_media, dtype=float)
        for c in range(len(channels)):
            c_max = np.max(working_media[:, :, c])
            if c_max > 0:
                media_scaled[:, :, c] = working_media[:, :, c] / c_max

        # Apply Adstock & Saturation
        time_steps, n_geos, n_channels = media_scaled.shape
        adstocked = np.zeros_like(media_scaled)

        for t in range(time_steps):
            if t == 0:
                adstocked[t] = media_scaled[t]
            else:
                adstocked[t] = media_scaled[t] + (alpha * adstocked[t-1])

        transformed = (adstocked ** slope) / ((adstocked ** slope) + (ec ** slope))

        # Apply Beta Coefficients and Sum for May
        c_contribs = {}
        c_total = 0.0

        for c_idx, channel in enumerate(channels):
            channel_contrib = 0.0
            for g_idx in range(n_geos):
                effect = beta[g_idx, c_idx] * transformed[:, g_idx, c_idx]
                channel_contrib += effect[may_mask].sum()

            c_contribs[str(channel)] = max(0, channel_contrib)
            c_total += max(0, channel_contrib)

        return c_contribs, c_total
    # ----------------------------------------------

    contributions = {}
    total_may_media = 0.0

    # 3. Process PAID Media
    try:
        media_raw = np.array(input_data.media)
        if hasattr(input_data, 'media_channels'):
            channels = np.array(input_data.media_channels)
        else:
            channels = idata.posterior.coords['media_channel'].values

        alpha = idata.posterior['alpha_m'].mean(dim=['chain', 'draw']).values
        ec = idata.posterior['ec_m'].mean(dim=['chain', 'draw']).values
        slope = idata.posterior['slope_m'].mean(dim=['chain', 'draw']).values

        if 'beta_gm' in idata.posterior:
            beta = idata.posterior['beta_gm'].mean(dim=['chain', 'draw']).values
        else:
            beta = idata.posterior['beta_m'].mean(dim=['chain', 'draw']).values
            beta = beta[np.newaxis, :]

        paid_contribs, paid_total = calculate_tensor_effects(media_raw, channels, alpha, ec, slope, beta)
        contributions.update(paid_contribs)
        total_may_media += paid_total
        print("✅ Paid media parameters extracted & processed.")
    except Exception as e:
        print(f"❌ Failed to process paid media: {e}")

    # 4. Process ORGANIC Media
    if 'organic_media_channel' in idata.posterior.coords:
        try:
            org_raw = np.array(input_data.organic_media)
            if hasattr(input_data, 'organic_media_channels'):
                org_channels = np.array(input_data.organic_media_channels)
            else:
                org_channels = idata.posterior.coords['organic_media_channel'].values

            alpha_om = idata.posterior['alpha_om'].mean(dim=['chain', 'draw']).values
            ec_om = idata.posterior['ec_om'].mean(dim=['chain', 'draw']).values
            slope_om = idata.posterior['slope_om'].mean(dim=['chain', 'draw']).values

            if 'beta_gom' in idata.posterior:
                beta_om = idata.posterior['beta_gom'].mean(dim=['chain', 'draw']).values
            else:
                beta_om = idata.posterior['beta_om'].mean(dim=['chain', 'draw']).values
                beta_om = beta_om[np.newaxis, :]

            org_contribs, org_total = calculate_tensor_effects(org_raw, org_channels, alpha_om, ec_om, slope_om, beta_om)

            # Label organic channels distinctly and add to the main dictionary
            for k, v in org_contribs.items():
                contributions[f"[Organic] {k}"] = v

            total_may_media += org_total
            print("✅ Organic media parameters extracted & processed.")
        except Exception as e:
            print(f"⚠️ Organic media found but failed to process: {e}")

    # 5. Calculate Percentages based on TOTAL media (Paid + Organic)
    if total_may_media == 0:
        print("⚠️ Total media contribution calculated as 0.")
        return None

    share_table = []
    for ch, vol in contributions.items():
        share_table.append({
            "Channel": ch,
            "May 2026 Contribution %": (vol / total_may_media) * 100
        })

    return pd.DataFrame(share_table).sort_values(by="May 2026 Contribution %", ascending=False)

# =====================================================================
# --- RUN PIPELINE ---
# =====================================================================

try:
    leads_path = "/content/model_leads.pkl"
    leads_shares = extract_may_shares_math(leads_path, data, "Leads Funnel")

    if leads_shares is not None:
        print("\n" + "🏆" * 25)
        print("🏆 MAY 2026 IN-SAMPLE CONTRIBUTION SHARES: LEADS")
        print("🏆" * 25)
        print(leads_shares.to_string(index=False, float_format="%.2f%%"))

except Exception as err:
    print(f"\n❌ Execution stopped for Leads Funnel: {err}")

try:
    apps_path = "/content/model_applications.pkl"
    apps_shares = extract_may_shares_math(apps_path, data, "Applications Funnel")

    if apps_shares is not None:
        print("\n" + "🏆" * 25)
        print("🏆 MAY 2026 IN-SAMPLE CONTRIBUTION SHARES: APPLICATIONS")
        print("🏆" * 25)
        print(apps_shares.to_string(index=False, float_format="%.2f%%"))

except Exception as err:
    print(f"\n❌ Execution stopped for Applications Funnel: {err}")
```
